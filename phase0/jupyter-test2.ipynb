```python
import pandas as pd

sample_data = pd.DataFrame({
    "model": ["Linear", "Random Forest", "Neural Net"],
    "accuracy": [0.72, 0.89, 0.94],
    "training_time": [0.1, 2.3, 45.6]
})
sample_data.to_csv("model_results.csv", index=False)
```

# 模型对比实验

```python
df = pd.read_csv("model_results.csv")
df
```

# Visualizagtion Comparison

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(df["model"]))
width = 0.35

fig, ax1 = plt.subplots(figsize=(8, 5))
ax2 = ax1.twinx()

bars1 = ax1.bar(x - width/2, df["accuracy"], width, label="Accuracy", color="#2a78d6")
bars2 = ax2.bar(x + width/2, df["training_time"], width, label="Training Time", color="#eb6834")
ax2.set_yscale('log')

ax1.set_xlabel("Model")
ax1.set_ylabel("Accuracy", color="#2a78d6")
ax2.set_ylabel("Training Time (s, log scale)", color="#eb6834")

ax1.set_xticks(x)
ax1.set_xticklabels(df["model"])

fig.tight_layout()
plt.show()
```
